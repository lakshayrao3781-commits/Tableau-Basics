Q8. What is the difference between a Worksheet and a Dashboard in Tableau?
Why is a dashboard more useful for decision-makers?

Worksheet vs. Dashboard: A worksheet contains a single view/chart, while a dashboard combines multiple worksheets onto a single screen for high-level monitoring.   

Why Useful for Decision-Makers: Provides an all-in-one executive snapshot, allowing leadership to interact with cross-filtered charts and spot key KPIs instantly.

Q9.  What is Tableau Prep, and why is it important in real-world data projects?
What is Tableau Prep, and why is it important in real-world data projects?
Then explain how you would:
Remove null values
Remove duplicate records
Change incorrect data types (Answer in steps, no tool access required.)

What is Tableau Prep: A dedicated ETL (Extract, Transform, Load) tool used to clean, shape, and combine raw data before visual analysis.   

Importance: Prevents data errors and saves time during analysis by standardizing dirty or unstructured data upfront.   

How to perform key cleaning operations:

Remove Null Values: Click on the field column header $\rightarrow$ filter out Null values.   

Remove Duplicate Records: Add a Clean Step $\rightarrow$ select fields $\rightarrow$ click option to deduplicate / aggregate.  

 Change Data Types: Click the data type icon above the column name (e.g., change string # icon to Date or Number).
